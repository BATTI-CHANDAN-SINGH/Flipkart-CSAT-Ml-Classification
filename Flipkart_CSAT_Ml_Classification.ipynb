## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

Answer Here.

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Null Hypothesis ($H_0$): There is no significant difference in the average CSAT score between newly onboarded agents (0 - 3 months tenure) and experienced agents (1+ years tenure).

**((Mu)experimented = (Mu)new)** Alternate Hypothesis ($H_1$): Experienced agents have a statistically significant higher average CSAT score than newly onboarded agents. **((Mu)experimented > (Mu)new)**

#### 2. Perform an appropriate statistical test.

````python
from scipy import stats

# Remove missing values
tenure_data = df.dropna(subset=['Tenure Bucket', 'CSAT Score'])

# Filter two distinct tenure groups
group_trainee = tenure_data[tenure_data['Tenure Bucket'] == 'On Job Training']['CSAT Score']
group_experienced = tenure_data[tenure_data['Tenure Bucket'] == '61-90']['CSAT Score']

# Perform Welch's t-test
t_stat, p_value = stats.ttest_ind(group_trainee, group_experienced, equal_var=False)

print(f"t-statistic: {t_stat:.4f}")
print(f"P-Value: {p_value:.4f}")

if p_value < 0.05:
    print("Result: Reject Null Hypothesis (H0) - There IS a significant difference!")
else:
    print("Result: Fail to Reject Null Hypothesis (H0) - NO significant difference found.")
````

##### Which statistical test have you done to obtain P-Value?

Two-Sample Independent $t $-Test (Welch's $t$-test).

##### Why did you choose the specific statistical test?

I chose a Two-Sample Independent $t$-Test because we are comparing the numerical averages (mean CSAT scores) of two separate categorical groups (On Job Training vs 61-90 days). Since the sample sizes are different (1,332 vs 873 entries), running Welch's $t$-test (equal_var=False) provides a robust, reliable $p$-value without assuming equal variance between the groups.

### Hypothetical Statement - 2

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

1. Null Hypothesis ($H_0$): There is no significant difference in average CSAT scores between trainees (On Job Training) and experienced agents (61-90 days tenure).
2. Alternate Hypothesis ($H_1$): There is a statistically significant difference in average CSAT scores between trainees (On Job Training) and experienced agents (61-90 days tenure)

#### 2. Perform an appropriate statistical test.

````python
from scipy import stats

# Convert Item_price to numeric
df_clean = df.copy()
df_clean['Item_price'] = pd.to_numeric(df_clean['Item_price'], errors='coerce')

# Drop missing values
price_csat_data = df_clean.dropna(subset=['Item_price', 'CSAT Score'])

# Split into Low CSAT (<= 3) and High CSAT (> 3)
low_csat_price = price_csat_data[price_csat_data['CSAT Score'] <= 3]['Item_price']
high_csat_price = price_csat_data[price_csat_data['CSAT Score'] > 3]['Item_price']

print(f"Low CSAT Sample Count: {len(low_csat_price)} | High CSAT Sample Count: {len(high_csat_price)}")

# Perform Welch's t-test
t_stat, p_value = stats.ttest_ind(low_csat_price, high_csat_price, equal_var=False)

print(f"t-statistic: {t_stat:.4f}")
print(f"P-Value: {p_value:.4f}")

if p_value < 0.05:
    print("Result: Reject Null Hypothesis (H0) - There IS a significant difference in item prices across satisfaction levels!")
else:
    print("Result: Fail to Reject Null Hypothesis (H0) - NO significant difference found.")
````

##### Which statistical test have you done to obtain P-Value?

Two-Sample Independent $t$-Test (Welch's $t$-test).

##### Why did you choose the specific statistical test?

I chose a Two-Sample Independent $t$-Test because we are comparing the numerical averages of a continuous metric (connected_handling_time) across two distinct satisfaction groups (Low CSAT vs High CSAT). Setting equal_var=False performs Welch's $t$-test, which safely accounts for any differences in sample sizes and variance between frustrated and happy customer interactions.

### Hypothetical Statement - 3

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

1. Null Hypothesis ($H_0$): Customer satisfaction levels (High CSAT vs Low CSAT) are independent of the support channel used (Inbound, Outbound, Chat, Email).
2. Alternate Hypothesis ($H_1$): Customer satisfaction levels are statistically dependent on the support channel used.

#### 2. Perform an appropriate statistical test.

````python
from scipy import stats
# Clean missing values
channel_csat_data = df.dropna(subset=['channel_name', 'CSAT Score']).copy()

# Categorize CSAT into 'Low CSAT' (<= 3) and 'High CSAT' (> 3)
channel_csat_data['CSAT_Group'] = channel_csat_data['CSAT Score'].apply(
    lambda x: 'Low CSAT' if x <= 3 else 'High CSAT'
)

# Create a contingency table (crosstab) between Channel and CSAT Group
contingency_table = pd.crosstab(channel_csat_data['channel_name'], channel_csat_data['CSAT_Group'])
print("Contingency Table:")
print(contingency_table)
print("-" * 50)

# Perform Chi-Square Test of Independence
chi2_stat, p_value, dof, expected = stats.chi2_contingency(contingency_table)

print(f"Chi-Square Statistic: {chi2_stat:.4f}")
print(f"Degrees of Freedom (dof): {dof}")
print(f"P-Value: {p_value:.4f}")

if p_value < 0.05:
    print("Result: Reject Null Hypothesis (H0) - Customer satisfaction is significantly DEPENDENT on the support channel!")
else:
    print("Result: Fail to Reject Null Hypothesis (H0) - Customer satisfaction is INDEPENDENT of the support channel.")
````

##### Which statistical test have you done to obtain P-Value?

Null Hypothesis ($H_0$): Customer satisfaction levels (High CSAT vs Low CSAT) are independent of the support channel used (Inbound, Outbound, Chat, Email).Alternate Hypothesis ($H_1$): Customer satisfaction levels are statistically dependent on the support channel used

##### Why did you choose the specific statistical test?

I chose a Chi-Square Test of Independence because we are examining the relationship between two categorical variables (channel_name and binned CSAT_Group). A Chi-Square test evaluates whether the distribution of high and low satisfaction ratings differs significantly across communication channels or if the observed counts match expected random proportions.

## ***6. Feature Engineering & Data Pre-processing***

````python
# Create a copy of the dataset to preserve original data
df_processed = df.copy()

# 1. Drop columns with extreme missing values (>70%)
cols_to_drop = ['connected_handling_time', 'Customer_City', 'Product_category', 'order_date_time', 'Item_price']
df_processed.drop(columns=[col for col in cols_to_drop if col in df_processed.columns], inplace=True)

# 2. Fill missing values in categorical/text columns
categorical_fill_unknown = ['Customer Remarks', 'category', 'Sub-category', 'Agent_name', 'Supervisor', 'Manager', 'Tenure Bucket', 'Agent Shift']
for col in categorical_fill_unknown:
    if col in df_processed.columns:
        df_processed[col] = df_processed[col].fillna('Unknown')

# 3. Drop rows where CSAT Score is missing (as it is our primary target metric)
df_processed.dropna(subset=['CSAT Score'], inplace=True)

# Verify missing values count
print("Remaining Missing Values:")
print(df_processed.isnull().sum()[df_processed.isnull().sum() > 0])
````

#### What all missing value imputation techniques have you used and why did you use those techniques?

1. **Column Dropping:** Columns like connected_handling_time ($100\%$ missing) and Customer_City / Product_category / Item_price ($>79\%$ missing) contained insufficient data to impute reliably without introducing extreme bias.

2. **Categorical Imputation ('Unknown'):** For categorical columns like Customer Remarks, category, and Sub-category, missing values were imputed with the string 'Unknown' to preserve row count while preventing data loss.
3. **Target Row Removal:** Rows missing the CSAT Score ($0.02\%$) were dropped because CSAT Score is our ground-truth target variable, and imputing target labels can corrupt model training.

### 2. Handling Outliers

````python
# Handling Outliers & Outlier treatments

# Select numerical features for outlier treatment (excluding target CSAT Score)
num_cols = df_processed.select_dtypes(include=[np.number]).columns.tolist()
if 'CSAT Score' in num_cols:
    num_cols.remove('CSAT Score')

# Apply Interquartile Range (IQR) Capping (Winsorization)
for col in num_cols:
    Q1 = df_processed[col].quantile(0.25)
    Q3 = df_processed[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Cap values outside bounds rather than dropping rows
    df_processed[col] = np.where(df_processed[col] < lower_bound, lower_bound, df_processed[col])
    df_processed[col] = np.where(df_processed[col] > upper_bound, upper_bound, df_processed[col])

print("Outlier treatment completed using IQR Capping.")
````

##### What all outlier treatment techniques have you used and why did you use those techniques?

1. **Technique Used:** Values lying below $\text{Q1} - 1.5 \times \text{IQR}$ or above $\text{Q3} + 1.5 \times \text{IQR}$ were capped at the lower and upper threshold values respectively.
2. **Why:** We used Capping instead of Trimming/Dropping because dropping outlier rows would lead to unnecessary data loss in support ticket logs where extreme handling/response times are legitimate long-tail occurrences. Capping reduces the leverage of extreme skewness while preserving the dataset size and overall data distribution

### 3. Categorical Encoding

````python
# Encode your categorical columns
from sklearn.preprocessing import LabelEncoder

# Identify categorical columns
categorical_cols = df_processed.select_dtypes(include=['object']).columns.tolist()

# 1. Label Encoding for high-cardinality and operational features
le = LabelEncoder()
for col in categorical_cols:
    if col in df_processed.columns:
        df_processed[col] = le.fit_transform(df_processed[col].astype(str))

print("Categorical encoding complete. Processed shape:", df_processed.shape)
````

#### What all categorical encoding techniques have you used & why did you use those techniques?

1. **Label Encoding:** Applied to convert text-based categories (e.g., channel_name, category, Sub-category, Tenure Bucket, Agent Shift) into numerical integer representations required by Scikit-Learn algorithms.

2. **Why Label Encoding over One-Hot Encoding?** Columns like Agent_name, category, and Sub-category contain high cardinality (dozens or hundreds of unique text values). Using One-Hot Encoding would create hundreds of sparse binary columns (the "curse of dimensionality"), memory bloat, and tree-based model inefficiency. Label Encoding retains a compact dataset shape while preparing all features cleanly for downstream modeling

### 4. Textual Data Preprocessing
(It's mandatory for textual dataset i.e., NLP, Sentiment Analysis, Text Clustering etc.)

#### 1. Expand Contraction

````python
# Expand
import re

# Dictionary mapping common English contractions to expanded form
contractions_dict = {
    "can't": "cannot", "won't": "will not", "n't": " not",
    "'re": " are", "'s": " is", "'d": " would",
    "'ll": " will", "'t": " not", "'ve": " have", "'m": " am"
}

def expand_contractions(text):
    text = str(text)
    for contraction, expansion in contractions_dict.items():
        text = re.sub(r'\b' + contraction + r'\b', expansion, text, flags=re.IGNORECASE)
    return text

# Apply expansion to text column
df_processed['Customer Remarks_clean'] = df_processed['Customer Remarks'].apply(expand_contractions)
print("Contractions expanded successfully.")
````

#### 2. Lower Casing

````python
# Lower Casing
# Convert all characters to lowercase
df_processed['Customer Remarks_clean'] = df_processed['Customer Remarks_clean'].str.lower()
print("Converted text to lowercase.")
````

#### 3. Removing Punctuations

````python
# Remove Punctuations
import string

# Remove all punctuation symbols using string translation
def remove_punctuation(text):
    return text.translate(str.maketrans('', '', string.punctuation))

df_processed['Customer Remarks_clean'] = df_processed['Customer Remarks_clean'].apply(remove_punctuation)
print("Punctuation removed.")
````

#### 4. Removing URLs & Removing words and digits contain digits.

````python
# Remove URLs & Remove words and digits contain digits
def remove_urls_and_digits(text):
    # Remove web URLs (http, https, www)
    text = re.sub(r'https?://\S+|www\.\S+', '', text)
    # Remove isolated numbers and alphanumeric tokens containing digits
    text = re.sub(r'\w*\d\w*', '', text)
    return text

df_processed['Customer Remarks_clean'] = df_processed['Customer Remarks_clean'].apply(remove_urls_and_digits)
print("URLs and digit-containing words removed.")
````

#### 5. Removing Stopwords & Removing White spaces

````python
# Remove Stopwords
import nltk
from nltk.corpus import stopwords

# Download stopwords list
nltk.download('stopwords', quiet=True)
stop_words = set(stopwords.words('english'))

# Remove English stopwords
def remove_stopwords(text):
    words = text.split()
    filtered_words = [word for word in words if word not in stop_words]
    return " ".join(filtered_words)

df_processed['Customer Remarks_clean'] = df_processed['Customer Remarks_clean'].apply(remove_stopwords)

# Remove extra whitespace spaces and newline characters
df_processed['Customer Remarks_clean'] = df_processed['Customer Remarks_clean'].apply(lambda x: re.sub(r'\s+', ' ', x).strip())
print("Stopwords and extra whitespace removed.")
````

#### 6. Rephrase Text

````python
# Rephrase Text
# Standardize common customer feedback shortcuts/abbreviations to full terms
rephrase_dict = {
    "deliv": "delivery", "delvy": "delivery", "shipment": "delivery",
    "dmg": "damaged", "pkg": "package", "cust": "customer", "exec": "agent"
}

def rephrase_text(text):
    words = text.split()
    rephrased = [rephrase_dict.get(w, w) for w in words]
    return " ".join(rephrased)

df_processed['Customer Remarks_clean'] = df_processed['Customer Remarks_clean'].apply(rephrase_text)
print("Text rephrasing applied.")
````

#### 7. Tokenization

````python
# Tokenization
# Split raw sentence string into individual token list
df_processed['tokens'] = df_processed['Customer Remarks_clean'].apply(lambda x: x.split())
print("Tokenization completed. Sample tokens:")
print(df_processed['tokens'].head(3))
````

#### 8. Text Normalization

````python
# Normalizing Text (i.e., Stemming, Lemmatization etc.)
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet', quiet=True)
lemmatizer = WordNetLemmatizer()

# Apply Lemmatization on tokens
def lemmatize_tokens(tokens):
    return [lemmatizer.lemmatize(token) for token in tokens]

df_processed['normalized_tokens'] = df_processed['tokens'].apply(lemmatize_tokens)
df_processed['Customer Remarks_final'] = df_processed['normalized_tokens'].apply(lambda x: " ".join(x))
print("Text normalization (Lemmatization) complete.")
````

##### Which text normalization technique have you used and why?

I used Lemmatization (WordNet Lemmatizer) over Stemming.Stemming relies on crude heuristic rules that cut off word endings (e.g., turning "delivery" into "deliv" or "badly" into "badl"), often producing non-dictionary words. In contrast, Lemmatization uses vocabulary and morphological analysis to transform words back to their meaningful dictionary base form (e.g., "better" $\rightarrow$ "good", "running" $\rightarrow$ "run"). This ensures that customer sentiment context is retained accurately for feature extraction

#### 9. Part of speech tagging

````python
# POS Taging
nltk.download('averaged_perceptron_tagger', quiet=True)
nltk.download('averaged_perceptron_tagger_eng', quiet=True)

# Generate Part-Of-Speech tags for each token
df_processed['pos_tags'] = df_processed['normalized_tokens'].apply(nltk.pos_tag)
print("POS Tagging completed. Sample POS output:")
print(df_processed['pos_tags'].head(2))
````

#### 10. Text Vectorization

````python
import pandas as pd
import numpy as np

# Convert column to string safely
raw_remarks = df_processed['Customer Remarks'].astype(str).str.strip()

# 1. Extract quantitative structural features from text
df_processed['remark_char_len'] = raw_remarks.apply(len)
df_processed['remark_word_count'] = raw_remarks.apply(lambda x: len(x.split()))
df_processed['has_remark'] = np.where(raw_remarks.isin(['', 'nan', 'Unknown', 'none']), 0, 1)

# 2. Extract presence matrix for models
text_features = df_processed[['remark_char_len', 'remark_word_count', 'has_remark']]

print("Text Feature Extraction Matrix Shape:", text_features.shape)
print("Extracted Features Summary:\n", text_features.describe().round(2))
````

````python
from sklearn.feature_selection import SelectKBest, f_classif
import warnings
warnings.filterwarnings('ignore')

# 1. Drop non-numeric, raw text, and target columns
cols_to_exclude = [
    'CSAT Score', 'is_high_csat', 'CSAT_Target', 'Customer Remarks',
    'Customer Remarks_clean', 'Customer Remarks_final', 'tokens',
    'normalized_tokens', 'pos_tags'
]

X = df_processed.drop(columns=[c for c in cols_to_exclude if c in df_processed.columns])
y = df_processed['is_high_csat']

# 2. Select numerical columns only
X_num = X.select_dtypes(include=['int64', 'float64', 'int32']).copy()

# 3. Drop constant features (zero variance) to prevent division-by-zero warnings
constant_cols = [col for col in X_num.columns if X_num[col].nunique() <= 1]
X_num.drop(columns=constant_cols, inplace=True)

# 4. Perform SelectKBest with ANOVA F-test
k_features = min(10, X_num.shape[1])
selector = SelectKBest(score_func=f_classif, k=k_features)
X_new = selector.fit_transform(X_num, y)

selected_cols = X_num.columns[selector.get_support()].tolist()
print("Top Selected Features:", selected_cols)
````

#### 2. Cross- Validation & Hyperparameter Tuning

````python
from sklearn.model_selection import RandomizedSearchCV

# Define Hyperparameter Space for XGBoost
param_dist_3 = {
    'n_estimators': [100, 200, 300],
    'max_depth': [3, 5, 7, 9],
    'learning_rate': [0.01, 0.05, 0.1, 0.2],
    'subsample': [0.7, 0.8, 0.9, 1.0],
    'colsample_bytree': [0.7, 0.8, 0.9, 1.0]
}

# Randomized Search CV Execution
search_3 = RandomizedSearchCV(
    estimator=XGBClassifier(random_state=42, eval_metric='logloss'),
    param_distributions=param_dist_3,
    n_iter=10,
    cv=5,
    scoring='f1',
    random_state=42,
    n_jobs=-1
)

search_3.fit(X_train_balanced, y_train_balanced)

# Predict with Tuned Model
best_model_3 = search_3.best_estimator_
y_pred_3_tuned = best_model_3.predict(X_test)
y_proba_3_tuned = best_model_3.predict_proba(X_test)[:, 1]

# Tuned Evaluation Chart
m3_tuned_metrics = {
    'Accuracy': accuracy_score(y_test, y_pred_3_tuned),
    'Precision': precision_score(y_test, y_pred_3_tuned),
    'Recall': recall_score(y_test, y_pred_3_tuned),
    'F1 Score': f1_score(y_test, y_pred_3_tuned),
    'ROC AUC': roc_auc_score(y_test, y_proba_3_tuned)
}

print("Best Parameters:", search_3.best_params_)
print("--- Tuned Model 3 Evaluation Chart ---")
print(pd.DataFrame([m3_tuned_metrics], index=['XGBoost Tuned']).round(4))
````
